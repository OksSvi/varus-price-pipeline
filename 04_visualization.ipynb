```python
%sql
-- Feeds the price-history chart on the dashboard.
-- Uses ROW_NUMBER() to keep only the latest snapshot per product per day,
-- since the pipeline can run more than once a day.
SELECT sku_code, product_name, brand, categories, scraped_date, scraped_at, price
FROM (
    SELECT 
        d.sku_code, d.product_name, d.brand, d.categories,
        f.scraped_date, f.scraped_at, f.price,
        ROW_NUMBER() OVER (
            PARTITION BY f.product_id, f.scraped_date 
            ORDER BY f.scraped_at DESC
        ) AS rn
    FROM gold_db.dim_products d
    JOIN gold_db.fct_prices f USING (product_id)
)
WHERE rn = 1
ORDER BY sku_code, scraped_date
```
