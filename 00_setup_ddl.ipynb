```python
%sql
-- Create the database that holds all pipeline tables (silver + gold layers)
CREATE DATABASE IF NOT EXISTS gold_db;
```

```python
%sql
-- Silver layer: cleaned, deduplicated product snapshots (one row per sku + scraped_at)
CREATE TABLE IF NOT EXISTS gold_db.silver_products (
    brand STRING,
    categories ARRAY<STRING>,
    category_path STRING,
    name STRING,
    price DOUBLE,
    scraped_at TIMESTAMP,
    sku STRING,
    url STRING,
    date DATE
)
USING DELTA
LOCATION 's3a://aws-scraping-test-oks/silver/products/';
```

```python
%sql
-- create dimensions table
CREATE TABLE IF NOT EXISTS gold_db.dim_products (
    product_id STRING,
    sku_code STRING,
    product_name STRING,
    brand STRING,
    category_path STRING,
    categories ARRAY<STRING>,
    product_url STRING,
    first_seen_date TIMESTAMP,
    last_seen_date TIMESTAMP
)
USING DELTA
LOCATION 's3a://aws-scraping-test-oks/gold/dim_products/';
```

```python
%sql
-- create fact table (price)
CREATE TABLE IF NOT EXISTS gold_db.fct_prices (
    price_id STRING,
    product_id STRING,
    scraped_date DATE,
    scraped_at TIMESTAMP,
    price DOUBLE
)
USING DELTA
LOCATION 's3a://aws-scraping-test-oks/gold/fct_prices/';
```
